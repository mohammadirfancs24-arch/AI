```python

import random
import math

def calculate_cost(state):
    """Calculates the number of mutually attacking pairs of queens."""
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_random_neighbor(state):
    """Generates a neighbor by moving one random queen to another row."""
    neighbor = list(state)
    col = random.randint(0, 7)
    row = random.randint(0, 7)
    while row == state[col]:
        row = random.randint(0, 7)
    neighbor[col] = row
    return neighbor

def print_board(state):
    """Prints the board in a visual grid representation."""
    for row in range(8):
        line = " ".join("Q" if state[col] == row else "." for col in range(8))
        print(line)
    print()

def simulated_annealing():
    # current <- initial state
    current = [random.randint(0, 7) for _ in range(8)]
    current_cost = calculate_cost(current)

    # T <- a large positive value
    T = 100.0
    decay_rate = 0.95

    steps = []
    steps.append(("Initial State", list(current), current_cost, T))

    # while T > 0 do
    while T > 0.01:
        if current_cost == 0:
            break

        # next <- a random neighbour of current
        neighbor = get_random_neighbor(current)
        neighbor_cost = calculate_cost(neighbor)

        # dE = current.cost - next.cost
        dE = current_cost - neighbor_cost

        # if dE > 0 then current <- next
        if dE > 0:
            current = neighbor
            current_cost = neighbor_cost
        else:
            # current <- next with probability p = e^(dE/T)
            # Note: dE is negative or zero, so e^(dE/T) works directly
            p = math.exp(dE / T)
            if random.random() < p:
                current = neighbor
                current_cost = neighbor_cost

        # decrease T
        T *= decay_rate

        # Record occasional progression steps
        if len(steps) < 5 or (current_cost < steps[-1][2] and len(steps) < 10):
            steps.append(("Iterating", list(current), current_cost, T))

    steps.append(("Final State", list(current), current_cost, T))
    return current, steps

solution, history = simulated_annealing()

# Showing only the most important steps
print("=== IMPORTANT SEARCH STEPS ===\n")
seen_costs = set()
for phase, state, cost, temp in history:
    # Only display meaningful state changes to keep output clean
    if cost not in seen_costs or phase in ["Initial State", "Final State"]:
        print(f"Phase: {phase} | Temp: {temp:.2f} | Conflicts (Cost): {cost} | State: {state}")
        seen_costs.add(cost)

print("\n=== FINAL BOARD ===")
print_board(solution)
```
